```python
df['GENERO'].unique()
```

```python
df['AREA CONOCIMIENTO'].unique()
```

```python
tabla_conteos = pd.crosstab(df['GENERO'], df['AREA CONOCIMIENTO'])
tabla_conteos
```

```python
total_estudiantes = tabla_conteos.sum().sum()
total_estudiantes
```

```python
tabla_prob_total = pd.crosstab(df['GENERO'], df['AREA CONOCIMIENTO'], normalize='all')
tabla_prob_total
```

```python
tabla_porcentaje_total = tabla_prob_total * 100
tabla_porcentaje_total
```

Si se selecciona al azar a un estudiante entre todos los matriculados de la región, la probabilidad de que sea hombre y estudie una carrera de Educación es de aproximadamente 0,0307 (3,07%).
